## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Task Type: Binary Classification 🏷️

The goal is to classify whether a given page/URL requires a content refresh (1 = Needs Refresh, 0 = Up to Date) based on historical traffic decay, freshness, and performance metrics.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target Variable: needs_refresh (Binary Proxy) 🎯

Since direct human labels might be unavailable, we define a proxy target based on traffic performance: a page is labeled as 1 (Needs Refresh) if its traffic in the recent period dropped by more than 30% compared to its baseline, otherwise 0.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: PR-AUC (Precision-Recall Area Under Curve) and F1-Score 📏

Why: The dataset is imbalanced (fewer pages actually need refresh). We want to maximize Precision (minimizing wasted editor effort on unnecessary updates) while maintaining high Recall (not missing pages that genuinely need updates).

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
%pip install pandas
```

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Define target proxy column
if "traffic_decay" in df.columns:
    df["target_needs_refresh"] = (df["traffic_decay"] > 0.30).astype(int)
else:
    df["target_needs_refresh"] = 1

# Display unit of analysis: 1 row = 1 unique URL / page
print(f"Total rows (pages): {len(df)}")
df[["url", "target_needs_refresh"]].head() if "url" in df.columns else df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML Beats Fixed Rules: 💡
Fixed rules (e.g., if age > 180 days then refresh) fail because traffic drop depends on multiple interacting non-linear features: seasonality, topic decay rate, competitor behavior, and impression changes. Machine Learning captures these multi-variable patterns to prioritize pages accurately rather than flagging static, arbitrary thresholds.
