## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

* **Validation Strategy:** Time-Aware (Temporal) Split.
* **Training Set ($T_0$):** Historical Search Console performance metrics prior to the cutoff date.
* **Test Set ($T_1$):** Subsequent, out-of-time data window used for realistic validation.
* **Anti-Leakage Justification:** A random split would leak future ranking dynamics, seasonal trends, and search engine changes into the training set. Using a temporal split mimics real-world deployment, evaluating how well the model predicts future opportunity volumes using past data.

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

* **Where the Model Fails:**
  * **Non-Linear CTR Decay:** Linear Regression assumes a strictly linear relationship with position. In reality, CTR drops exponentially after position 3, leading to overestimating opportunity volumes for positions beyond page 1.
  * **Impression Outliers:** Pages targeting broad or generic terms with massive impressions distort the linear prediction, creating inflated gap scores.
* **Actionable Recommendation:** Future iterations should incorporate non-linear models (such as Random Forest or XGBoost) or apply log-transformations to feature representations to better mirror real-world SERP behavior.
