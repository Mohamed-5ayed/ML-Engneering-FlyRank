## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

* **Intended Use:** This playbook serves strictly as a **decision-support tool** to help content and SEO teams prioritize manual page refreshes and title optimizations.
* **Operational Limits:**
  * **Non-Causal Outputs:** Model scores reflect statistical correlation ($\rho$), not guaranteed traffic growth.
  * **Static Context:** The model cannot anticipate sudden algorithm updates, SERP feature layout changes, or real-time competitor campaigns.
  * **Human-in-the-Loop Mandate:** Every generated recommendation requires human review prior to CMS deployment.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

To protect site integrity and search indexing, strict boundaries define what can be automated versus what requires mandatory human review.

* **Mandatory Human Review Triggers:**
  * Title/Meta description edits for top-performing URLs (Positions 1–5).
  * Major structural heading shifts or content section updates.

* **The No-Go List (STRICTLY PROHIBITED FROM AUTOMATION):**
  * 🛑 **URL Structural Changes & 301 Redirects:** High risk of breaking link equity.
  * 🛑 **Robots.txt & Canonical Tag Edits:** Misconfigurations can trigger site-wide de-indexing.
  * 🛑 **Automated Content Deletion:** Risks removing valuable semantic keyword coverage.
